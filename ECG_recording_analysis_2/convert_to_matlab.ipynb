```python

from pathlib import Path
import neo
import scipy.io

# Carpeta base
folder = Path('')          # por ejemplo Path('/ruta/a/tus/datos')
exact_name = 'experiments/VNS-008'

input_dir = folder / exact_name
output_dir = folder / exact_name

# Recorre todos los archivos .smr en la carpeta
for smr_file in input_dir.glob('*.smr'):
    print(f'Procesando: {smr_file.name}')

    # Leer el archivo SMR
    reader = neo.Spike2IO(filename=str(smr_file))
    data = reader.read()[0]

    # Extraer la señal
    signal = data.segments[0].analogsignals[0]

    # Convertir a formato guardable
    signal_array = signal.magnitude
    sampling_rate = signal.sampling_rate.magnitude

    # Nombre del archivo de salida (.mat)
    mat_file = output_dir / (smr_file.stem + '.mat')

    # Guardar
    scipy.io.savemat(mat_file, {
        'signal': signal_array,
        'sampling_rate': sampling_rate
    })

print('Todos los archivos han sido procesados.')
```
